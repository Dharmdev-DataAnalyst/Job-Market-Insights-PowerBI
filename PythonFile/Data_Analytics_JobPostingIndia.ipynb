```python
Y=dff['base salary']
X=dff.drop('base salary',axis=1)
```

```python
#Instead using LabelEncoder i used one hot encoding method to convert
X=pd.get_dummies(X,drop_first=True)
```

```python
Y
```

```python
X
```

```python
#Split the data in train and test part
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(X,Y,test_size=0.20,random_state=42)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(x_train,y_train)
```

```python
pred=model.predict(x_test)
pred
```

```python
from sklearn.metrics import *
res=pd.DataFrame(columns=['Actual','Predicted'])
res['Actual']=y_test
res['Predicted']=pred
```

```python
res
```

```python
r2=round(r2_score(y_test,pred),1)
r2
```

```python
mae=mean_absolute_error(y_test,pred)
mae
```

```python
mse=mean_squared_error(y_test,pred)
mse
```

```python
rmse=np.sqrt(mse)
rmse
```

Here R2 score is 0.5 means the model explains 50% of the variation in the base salary.This can also consider in moderate performance by model.

The MSE of 21525544155.93 is difficult to interpret by itself because it is measured in squared salary units,thats why after calculating RMSE that is 146715.86is indicating that model's prediction off by 1,46,715 lakhs on average.

it can be made more accurate by using Random Forest Regressor model.


```python
from sklearn.ensemble import RandomForestRegressor
model1=RandomForestRegressor(n_estimators=100,random_state=42)
model1.fit(x_train,y_train)
```

```python
ypred=model1.predict(x_test)
```

```python
ypred
```

```python
ress=pd.DataFrame(columns=['Actual','Predicted'])
ress['Actual']=y_test
ress['Predicted']=ypred
res
```

```python
r2score=round(r2_score(y_test,ypred),1)
r2score
```

```python
mae1=mean_absolute_error(y_test,ypred)
mae1
```

```python
mse1=mean_squared_error(y_test,ypred)
mse1
```

```python
rmse1=np.sqrt(mse1)
rmse1
```

The Random Forest Regressor outperformed the Linear Regression model by capturing nonlinear relationships in the data. It achieved a higher r2 score and lower prediction errors like(mae,rmse), indicating better predictive performance for estimating base salary
